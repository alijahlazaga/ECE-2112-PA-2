```python
import numpy as np
```

```python
# A. REPRODUCIBLE NORMALIZATION PROBLEM
```

```python
np.random.seed(2112) #Starting point for generazing randomized numbers for the array
X = np.random.randint(10, 101, size=(5, 5)) #Generates randomized integers within a range; the range being low, high, and size.
X
```

```python
X.mean() #Calculates the mean of the array X.
```

```python
X.std() #Calculates the standard deviation of the array X.
```

```python
X_normalized = (X - X.mean())/X.std() # Normalizes the data with the use of the Z-score formula.
X_normalized
```

```python
X_normalized.mean()#Calculates the mean of array X_normalized.
```

```python
X_normalized.std() #Calculates the standard deviation of array X_normalized.
```

```python
np.save("X_normalized.npy", X_normalized) #Saves the created array
```

```python
# B. CUBES DIVISIBLE BY 4 PROBLEM
```

```python
N = np.arange(1, 101) #Generates an array with the specified starting point and end point.
N
```

```python
C = (N ** 3).reshape(10, 10) #Cubes the array N and reshapes into 10 rows and 10 columns, it then renames it to C.
C
```

```python
div_by_4 = C[C % 4 == 0] #Returns the remainder of dividing the data inside the array by 4. It then keeps the values that stay true to the condition; the condition being equals to 0.
div_by_4
```

```python
np.save("div_by_4.npy", div_by_4) #Saves the created array
```

```python
# C. ABOVE-MEAN SQUARES PROBLEM
```

```python
S = np.arange(1, 37).reshape(6, 6) #Generates an array with randomized numbers with a specified starting point and end point. It then reshapes the array with 6 rows and 6 columns.
S
```

```python
S_mean = S.mean() #Calculates the mean of array S.
S_mean
```

```python
above_mean = S[S > S_mean] #Retrieves the numbers above the mean.
above_mean
```

```python
np.save("above_mean.npy", above_mean)
```
